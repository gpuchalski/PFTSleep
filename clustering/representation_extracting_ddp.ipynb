```python
# -------------------------
# 7) Extract + cache latents per encoder
# -------------------------
from pathlib import Path
import json
def cache_tag(
    encoder_name: str,
    num_files: int,
    frequency: int,
    win_length: int,
    hop_length: int,
    max_seq_len_sec: int,
) -> str:
    return (
        f"{encoder_name}"
        f"__files{num_files}"
        f"__freq{frequency}"
        f"__win{win_length}"
        f"__hop{hop_length}"
        f"__max{max_seq_len_sec}"
    )

def extract_latents_cached(
    encoder_name: str,
    encoder: torch.nn.Module,
    loader: DataLoader,
    cache_dir: Path,
    window_size_sec: float,
    device: str = "cpu",
):
    tag = cache_tag(
        encoder_name=encoder_name,
        num_files=num_files,
        frequency=frequency,
        win_length=win_length,
        hop_length=hop_length,
        max_seq_len_sec=max_seq_len_sec,
    )

    Z_path = cache_dir / f"Z__{tag}.npz"
    night_id_path = cache_dir / f"night_id__{tag}.npy"
    time_idx_path = cache_dir / f"time_idx__{tag}.npy"
    zarr_file_idx_path = cache_dir / f"zarr_file_idx__{tag}.npy"
    windows_start_idx_path = cache_dir / f"windows_start_idx__{tag}.npy"
    zarr_files_list_path = cache_dir / f"zarr_files_list__{tag}.json"

    if (Z_path.exists() and night_id_path.exists() and time_idx_path.exists() 
        and zarr_file_idx_path.exists() and windows_start_idx_path.exists()
        and zarr_files_list_path.exists()):
        print(f"[{encoder_name}] Loading cached latents:")
        Z = np.load(Z_path)['Z']
        night_id = np.load(night_id_path)
        time_idx = np.load(time_idx_path)
        zarr_file_idx_arr = np.load(zarr_file_idx_path)
        windows_start_idx_arr = np.load(windows_start_idx_path)
        with open(zarr_files_list_path, 'r') as f:
            zarr_files_list = json.load(f)
        
        print(f"  Z: {Z.shape}, nights: {len(np.unique(night_id))}, zarr files: {len(zarr_files_list)}")
        return Z, night_id, time_idx, zarr_file_idx_arr, windows_start_idx_arr

    print(f"[{encoder_name}] No cache found. Extracting latents on GPU (one-time cost)...")
    encoder = encoder.to(device)
    encoder.eval()

    latent_list = []
    night_ids = []
    time_idxs = []
    zarr_file_idxs = []
    windows_start_idxs = []

    with torch.no_grad():
        for batch_idx, batch in enumerate(tqdm(loader, desc=f"Encoding ({encoder_name}) on {device.upper()}", unit="batch")):
            if device == "cuda":
                torch.cuda.empty_cache()
            
            x = batch[0].to(device)
            sequence_padding_mask = batch[2].to(device)
            metadata = batch[3]

            with torch.cuda.amp.autocast(enabled=(device=="cuda")):
                z_latent = encoder(x, sequence_padding_mask=sequence_padding_mask)
            
            z_latent = z_latent.squeeze(0)
            z_latent = z_latent.permute(0, 2, 1)
            z_latent = z_latent.mean(dim=0)

            z_np = z_latent.detach().cpu().numpy()
            latent_list.append(z_np)

            n_windows = z_np.shape[0]
            
            file_path = metadata["file"][0]
            numeric_id = file_to_numeric_id[file_path]
            start_idx = metadata["start_idx"][0]
    
            night_ids.append(np.full(n_windows, batch_idx, dtype=int))
            time_idxs.append(np.arange(n_windows, dtype=float) * window_size_sec)
            zarr_file_idxs.append(np.full(n_windows, numeric_id, dtype=np.int32))
            windows_start_idxs.append(np.full(n_windows, start_idx))
            
            del x, sequence_padding_mask, z_latent
            if device == "cuda":
                torch.cuda.empty_cache()
    print("ABOUT TO VSTACK")
    Z = np.vstack(latent_list)
    print("VSTACK DONE")
    night_id = np.concatenate(night_ids)
    time_idx = np.concatenate(time_idxs)
    zarr_file_idx_arr = np.concatenate(zarr_file_idxs)
    window_start_idx_arr = np.concatenate(windows_start_idxs)

    print("ABOUT TO SAVE COMPRESSED")
    np.savez_compressed(Z_path, Z=Z)
    print("SAVED COMPRESSED")
    np.save(night_id_path, night_id)
    np.save(time_idx_path, time_idx)
    np.save(zarr_file_idx_path, zarr_file_idx_arr)
    np.save(windows_start_idx_path, window_start_idx_arr)
    
    zarr_files_str_list = [str(p) for p in zarr_files]
    with open(zarr_files_list_path, 'w') as f:
        json.dump(zarr_files_str_list, f)

    print(f"[{encoder_name}] Saved cache:")
    print(f"  {Z_path.name}  shape={Z.shape}")
    print(f"  Saved metadata for {len(zarr_files_str_list)} zarr files")

    return Z, night_id, time_idx, zarr_file_idx_arr, window_start_idx_arr

print(f"\n🚀 Starting latent extraction on {device.upper()}...\n")
LATENTS = {}
for enc_name, spec in ENCODER_REGISTRY.items():
    model = spec["build"]()
    model = spec["load"](model)
    Z, night_id, time_idx, zarr_file_idx, windows_start_idx = extract_latents_cached(
        encoder_name=enc_name,
        encoder=model,
        loader=loader,
        cache_dir=cache_dir,
        window_size_sec=window_size_sec,
        device=device,
    )
    LATENTS[enc_name] = {"Z": Z, "night_id": night_id, "time_idx": time_idx, "zarr_file_idx": zarr_file_idx, "windows_start_idx": windows_start_idx}

Z = LATENTS["PFTSleep"]["Z"]
zarr_file_idx = LATENTS["PFTSleep"]["zarr_file_idx"]
night_id = LATENTS["PFTSleep"]["night_id"]
time_idx = LATENTS["PFTSleep"]["time_idx"]
windows_start_idx = LATENTS["PFTSleep"]["windows_start_idx"]

print("\n✅ Latent extraction complete!")
print(f"Final latent matrix Z: {Z.shape}")
print(f"Zarr File Index Shape: {zarr_file_idx.shape}")
print(f"Windows Start Index Shape: {windows_start_idx.shape}")
print(f"Data from {len(np.unique(zarr_file_idx))} unique zarr files")
```

### Files Saved (per encoder)
Pattern: `{variable}__{encoder_name}__files{num_files}__freq{frequency}__win{win_length}__hop{hop_length}__max{max_seq_len_sec}.npy`

1. **Z__{tag}.npy** - Latent representations [N_windows, 512]

3. **time_idx__{tag}.npy** - Time index in seconds for each window
4. **zarr_file_idx__{tag}.npy** - Numeric zarr file ID for each window (maps to zarr_id_map.json)
5. **windows_start_idx__{tag}.npy** - Start index in original zarr for each window
6. **zarr_files_list__{tag}.npy** - List of zarr file paths (for mapping back to source)
7. **zarr_id_map.json** - Mapping from zarr UID to numeric ID
